# VMLU Dataset Discovery and Summary
This notebook explores the VMLU dataset. We will:
1. Load the dataset files (`dev.jsonl`, `test.jsonl`, `valid.jsonl`).
2. Calculate how many questions exist for each task/subject.
3. Print a sample question for each task.


```python
import json
import os
from collections import defaultdict

data_dir = "vmlu_mqa_v1.5"
splits = ["dev", "test", "valid"]
dataset = []

for split in splits:
    file_path = os.path.join(data_dir, f"{split}.jsonl")
    if os.path.exists(file_path):
        with open(file_path, "r", encoding="utf-8") as f:
            for line in f:
                item = json.loads(line)
                item["split"] = split
                dataset.append(item)

print(f"Total questions loaded: {len(dataset)}")
```

## Task and Subject Analysis
The `id` field in the dataset has a prefix (e.g., `28-0001`). We will extract the task/subject ID (the part before the hyphen) and count the number of questions for each task.


```python
task_counts = defaultdict(int)
sample_questions = {}

for item in dataset:
    # Assuming ID is like "28-0001" or similar
    task_id = item["id"].split("-")[0]
    task_counts[task_id] += 1
    
    if task_id not in sample_questions:
        sample_questions[task_id] = item

print(f"{'Task ID':<10} | {'Question Count':<15}")
print("-" * 27)
for task_id in sorted(task_counts.keys()):
    print(f"{task_id:<10} | {task_counts[task_id]:<15}")
```

## Sample Questions for Each Task
Let's print one sample question with its choices and answer for each task to understand the subjects.


```python
for task_id in sorted(sample_questions.keys()):
    sample = sample_questions[task_id]
    print(f"--- Task ID: {task_id} ---")
    print(f"Question: {sample['question']}")
    if 'choices' in sample:
        print("Choices:")
        for choice in sample['choices']:
            print(f"  {choice}")
    if 'answer' in sample:
        print(f"Answer: {sample['answer']}")
    print("\n")
```
